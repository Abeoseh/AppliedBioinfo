```python
import pandas as pd
import csv
```

```python
file = 'GCF_977971745.1_Znev_1.0_genomic.gff/GCF_977971745.1_Znev_1.0_genomic.gff'
# df = pd.read_table('GCF_977971745.1_Znev_1.0_genomic.gff/GCF_977971745.1_Znev_1.0_genomic.gff', comment='#', sep=r'[\t;]', quoting=csv.QUOTE_NONE, engine='python')

df = pd.read_csv(file, sep='\t', comment='#', header=0, names=["seqid","source","type","start","end","score","strand","phase","attributes"])
```

```python
df
```

```python
genes = df[df.type == "gene"]
```

```python
genes.sort_values(by=["seqid", "start"], inplace = True)
genes = genes.reset_index(drop=True)
```

start position of next gene - end position of previous gene - 1


`seqid` identifies a reference sequence, which may be a contig, scaffold, chromosome, or other sequence unit.

```python
grouped_seqid = genes.groupby("seqid")
```

```python
for i in grouped_seqid:
    print(i)
    print("------------")
```

```python
distances = []
distance_info = {}
for group in grouped_seqid:
    # print(group[0])
    df = group[1]
    distance_info[group[0]] = []
    for i in range(min(df.index), max(df.index), 2):
        if df.loc[i+1,"start"] > df.loc[i, "end"]:
            distances.append(int(df.loc[i+1,"start"]) - int(df.loc[i, "end"]) - 1)
            distance_info[group[0]].append((f"start next: {int(df.loc[i+1,"start"])} end previous: {int(df.loc[i, "end"])}"))
        else:
            # print(f"for gene {group[0]} start of next gene: {df.loc[i+1,"start"]} is within the previous gene: {df.loc[i, "end"]} ")
            distances.append(0)
            distance_info[group[0]].append((f"start next: {int(df.loc[i+1,"start"])} end previous: {int(df.loc[i, "end"])}"))



        if i+2 in df.index:
            if (df.loc[i+2,"start"] > df.loc[i+1, "end"]):
                distances.append(int(df.loc[i+2,"start"]) - int(df.loc[i+1, "end"]) - 1)
                distance_info[group[0]].append((f"start next: {int(df.loc[i+2,"start"])} end previous: {int(df.loc[i+1, "end"])}"))

            else:
                # print(f"for gene {group[0]} start of next gene: {df.loc[i+2,"start"]} is within the previous gene end: {df.loc[i+1, "end"]} ")
                distances.append(0)
                distance_info[group[0]].append((f"start next: {int(df.loc[i+2,"start"])} end previous: {int(df.loc[i+1, "end"])}"))


        
    
```

```python
print(f"average distance: {sum(distances)/len(distances)}")
print(f"median distance: {sorted(distances)[len(distances)//2]}")
print(f"range not including 0s: min {min([distance for distance in distances if distance != 0])}, max {max(distances)}")
```
